```python
import os
import requests
from dotenv import load_dotenv

load_dotenv()
```

```python
url = "https://crewai-deploy-7d5ca2b5-ab31-45e6-bf77-ca7ea-85d0f46c.crewai.com"

token = os.getenv("BEARER_TOKEN")

headers = {
    "Authorization": f"Bearer {token}",
    "Content-Type": "application/json"
}

if not token:
    print("Nao achei BEARER_TOKEN no .env")
    print("Pasta onde o notebook esta rodando:", os.getcwd())
    print("Tem um .env aqui?", os.path.exists(".env"))
else:
    print("Token carregado do .env! Termina em:", token[-4:])
```

```python
response = requests.get(f"{url}/inputs", headers=headers)
if response.status_code == 200:
    print(response.json())
else:
    print(f"Erro {response.status_code}: {response.text}")
```

```python
body = {
    "inputs":{
        "topic": "CrewAI",
        "current_year": 2026
    }
}

kickoff_id = None
response = requests.post(f"{url}/kickoff", headers=headers, json=body)
if response.status_code == 200:
    kickoff_id = response.json()["kickoff_id"]
    print("Comecou! ID:", kickoff_id)
else:
    print(f"Erro {response.status_code}: {response.text}")
```

```python
response = requests.get(f"{url}/status/{kickoff_id}", headers=headers)
if response.status_code == 200:
    dados = response.json()
    estado = dados["state"]
    print("Estado:", estado)
    if estado == "SUCCESS":
        print()
        print(dados["result"])
    elif estado == "STARTED":
        print("Ainda pensando. Espere uns segundos e rode esta celula de novo.")
    else:
        print(dados["status"])
else:
    print(f"Erro {response.status_code}: {response.text}")
```
